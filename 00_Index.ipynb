### 🧩 One-line map of the whole subject

- **State-space models** describe *how the world moves* (dynamics + measurements + noise).  →  [01](01_Continuous_State_Space_Models.ipynb)–[06](06_Stochastic_Processes_and_Propagating_Uncertainty.ipynb)
- **Sequential probabilistic inference** = *predict, then correct* with the conditional mean.  →  [07](07_Sequential_Probabilistic_Inference_Six_Steps.ipynb)
- The **Kalman filter** is SPI made exact for **linear-Gaussian** systems.  →  [08](08_Deriving_the_Linear_Kalman_Filter.ipynb)–[11](11_Target_Tracking_IMM_AlphaBetaGamma.ipynb)
- The **EKF/SPKF** extend it to **nonlinear** systems (linearize vs. sigma-points).  →  [12](12_Extended_Kalman_Filter.ipynb)–[14](14_Parameter_Estimation_Joint_Dual.ipynb)
- **Particle filters** handle **arbitrary non-Gaussian / multimodal** posteriors via Monte-Carlo.  →  [15](15_Bayesian_Recursion_MonteCarlo_Importance_Sampling.ipynb)–[17](17_Particle_Filter_Navigation_Application.ipynb)

→ Every method fills the **same predict/correct skeleton**; they differ only in *how* they push a probability distribution through the model.
