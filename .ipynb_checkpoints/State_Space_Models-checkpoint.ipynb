### 🧩 Example of a Continuous-Time State-Space Model

- Representation of the dynamics of an $n^{\text{th}}$-order system as a first-order differential equation in an $n$-vector called the **state**.  
  → $n$ coupled first-order equations.

- **Classic example:** Second-order equation of motion  

  $$
  m\ddot{z}(t) = u(t) - b\dot{z}(t) - k z(t)
  $$

  or equivalently  

  $$
  \ddot{z}(t) = \frac{u(t) - b\dot{z}(t) - k z(t)}{m}
  $$

  where $\ddot{z}(t) = \frac{d^2 z(t)}{dt^2}$ and $\dot{z}(t) = \frac{dz(t)}{dt}$, etc.

- Define a (non-unique) state vector  

  $$
  x(t) = 
  \begin{bmatrix}
  z(t) \\[4pt]
  \dot{z}(t)
  \end{bmatrix},
  \quad
  \dot{x}(t) = 
  \begin{bmatrix}
  \dot{z}(t) \\[4pt]
  \ddot{z}(t)
  \end{bmatrix}
  =
  \begin{bmatrix}
  0 & 1 \\[4pt]
  -\dfrac{k}{m} & -\dfrac{b}{m}
  \end{bmatrix}
  \begin{bmatrix}
  z(t) \\[4pt]
  \dot{z}(t)
  \end{bmatrix}
  +
  \begin{bmatrix}
  0 \\[4pt]
  \dfrac{1}{m}
  \end{bmatrix}
  u(t)
  $$

---

### 🧩 Example in Continuous-Time State-Space Form

- So far, we have  

  $$
  \dot{x}(t) =
  \begin{bmatrix}
  \dot{z}(t) \\[4pt]
  \ddot{z}(t)
  \end{bmatrix}
  =
  \begin{bmatrix}
  0 & 1 \\[4pt]
  -\dfrac{k}{m} & -\dfrac{b}{m}
  \end{bmatrix}
  \begin{bmatrix}
  z(t) \\[4pt]
  \dot{z}(t)
  \end{bmatrix}
  +
  \begin{bmatrix}
  0 \\[4pt]
  \dfrac{1}{m}
  \end{bmatrix}
  u(t)
  $$

- We can write this as $\dot{x}(t) = A x(t) + B u(t)$, where $A$ and $B$ are constant matrices.

  $$
  A =
  \begin{bmatrix}
  0 & 1 \\[4pt]
  -\dfrac{k}{m} & -\dfrac{b}{m}
  \end{bmatrix},
  \quad
  B =
  \begin{bmatrix}
  0 \\[4pt]
  \dfrac{1}{m}
  \end{bmatrix}
  $$

- Complete the model by computing $z(t) = C x(t) + D u(t)$, where $C$ and $D$ are constant matrices.

  $$
  C = [1 \quad 0], \quad D = [0]
  $$

---

### 🧩 Standard State-Space Model Form

- Standard form for **continuous-time linear** state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

  where $u(t)$ is the known input, $x(t)$ is the state, and $A, B, C, D$ are constant matrices.

- Convenient way to express dynamics (matrix format great for computers).

- The first equation is called the **state equation** or **process equation**.  
  → It evolves over time, summarizing the *dynamics* of the model.

- The second equation is called the **output equation** or **measurement equation**.  
  → A static linear combination of variables known at time $t$.


### 🧩 What is the System State Vector?

- Standard form for continuous-time linear state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

- We think of the system **state** at time $t_0$ as a minimum set of information at $t_0$ that, together with input $u(t),\ t \ge t_0$, uniquely determines system behavior for all $t \ge t_0$.
  - State variables provide access to what is going on **inside** the system.

- The state has dimensions $x(t) \in \mathbb{R}^n$, so $A \in \mathbb{R}^{n \times n}$ and $w(t) \in \mathbb{R}^n$.

- In principle, we can solve the state equation numerically:

  $$
  x(t) = x(0) + \int_{0}^{t} \big[ A x(\tau) + B u(\tau) + w(\tau) \big] d\tau
  $$

  but it is usually more convenient to keep the equation in **differential form**.

---

### 🧩 What are the Output and Inputs?

- Standard form for continuous-time linear state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

- The model output is $z(t) \in \mathbb{R}^m$. This is the **measurement** we make.

- There are three input signals to the model: $u(t)$, $w(t)$, and $v(t)$.

- We consider $u(t) \in \mathbb{R}^r$ to be a **deterministic** input;  
  that is, we assume that we know its value exactly at all times.

  - Based on the size of $u(t)$, we infer that $B \in \mathbb{R}^{n \times r}$.
  - The deterministic input forces $x(t)$ to evolve over time in different ways, depending on its values.
  - It also influences the output via the $D$ term.

---

### 🧩 What are the Random (Stochastic) Inputs?

- Standard form for continuous-time linear state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

- The model has two **random input** signals: $w(t)$ and $v(t)$.

- They are random in the sense that we assume we never know their values exactly.

  - We have already seen that $w(t) \in \mathbb{R}^n$;  
    if $z(t) \in \mathbb{R}^m$, then $v(t) \in \mathbb{R}^m$ also.

- The $w(t)$ signal is **process noise**.  
  → It affects the dynamics of the model by making direct changes to the evolution of $x(t)$.

- The $v(t)$ signal is **sensor noise**.  
  → It does not affect the dynamics of the model; it affects only the measurement $z(t)$.

- Systems having noise inputs $w(t)$ and $v(t)$ are considered in detail in week 3.


### 🧩 What are the Matrices Called?

- Standard form for continuous-time linear state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

- $A \in \mathbb{R}^{n \times n}$ is the **system matrix**.  
  → It models the evolution of the state in the absence of input.

- $B \in \mathbb{R}^{n \times r}$ is the **input matrix**.  
  → It defines how linear combinations of $u(t)$ impact the evolution of the state.

- $C \in \mathbb{R}^{m \times n}$ is the **output matrix**.  
  → It defines how the output depends on linear combinations of states.

- $D \in \mathbb{R}^{m \times r}$ is the **feedforward (or feedthrough) matrix**.  
  → It models how the output depends on linear combinations of the input (instantaneously).

- Time-varying systems have $A, B, C, D$ that change with time.

---

### 🧩 Why Do You Need to Know About Them?

- It can be of vital interest to know the **state** of the system,  
  but we usually cannot measure it directly.

- Instead, we can measure $u(t)$ and $z(t)$, and although we cannot measure the random signals $w(t)$ and $v(t)$,  
  we can model some of their critical attributes.

- This allows us to make methods to **estimate** $x(t)$ based on what we measure and what we model.

- **Kalman Filters** are (in some cases optimal) **estimators** of $x(t)$.

  - The derivation and implementation of the KF depend on modeling the system in state-space form  
    and having knowledge of the model matrices $A, B, C,$ and $D$.
  - This is why we care!


### 🧩 The Target-Tracking Application

- One application of KFs is for **target tracking**.  
  We seek to predict the **future location** of a dynamic system (the target) based on KF estimates and measurements.

- Ideally, we would know all matrices and signals of its continuous-time linear state-space model:

  $$
  \dot{x}(t) = A x(t) + B u(t) + w(t)
  $$
  $$
  z(t) = C x(t) + D u(t) + v(t)
  $$

- However, in the target-tracking application:
  - We don’t generally know the input signal $u(t)$.
  - We probably don’t know the state-space matrices that describe the target’s dynamics very well either.

- So, we approximate target dynamics based on some **assumed behaviors**, e.g.:
  - The target is generally **stationary**, or
  - Tends to move in a **straight line**, or
  - Moves in **circles**, etc.

---

### 🧩 Example: The Nearly-Constant-Position (NCP) Model

- Consider a **relatively immobile object** that we would like to track, which gets bumped around by unknown forces.

- For motion in 2-D, define the state vector as:

  $$
  x(t) =
  \begin{bmatrix}
  \xi(t) \\[4pt]
  \eta(t)
  \end{bmatrix}
  $$

  where $\xi(t)$ is the **x-coordinate** and $\eta(t)$ is the **y-coordinate** of position.

- The **state equation** is:

  $$
  \dot{x}(t) = 0 x(t) + w(t)
  $$

  where $w(t)$ is an **unknown (random) process-noise input**, unlike the deterministic $u(t)$.

  - $A = 0_{2\times2}$
  - $B = 0$
  - The size of $B$ depends on unknown $u(t)$, so it isn’t well-defined.

- One possible **output equation** is:

  $$
  z(t) = x(t) + v(t)
  $$

  where $v(t)$ is an unknown **sensor-noise input**.

  - $C = I_{2\times2}$
  - $D = 0$
  - The dimensions of $D$ are not well-defined either.

---

### 🧩 Example: The Nearly-Constant-Velocity (NCV) Model

- Now consider an object having **momentum** — its velocity is nearly constant but gets perturbed by external forces.

- Define the state and state equation as:

  $$
  x(t) =
  \begin{bmatrix}
  \xi(t) \\[4pt]
  \dot{\xi}(t) \\[4pt]
  \eta(t) \\[4pt]
  \dot{\eta}(t)
  \end{bmatrix},
  \quad
  \dot{x}(t) =
  \underbrace{
  \begin{bmatrix}
  0 & 1 & 0 & 0 \\[4pt]
  0 & 0 & 0 & 0 \\[4pt]
  0 & 0 & 0 & 1 \\[4pt]
  0 & 0 & 0 & 0
  \end{bmatrix}
  }_{A}
  x(t)
  +
  \underbrace{
  \begin{bmatrix}
  0 & 0 \\[4pt]
  1 & 0 \\[4pt]
  0 & 0 \\[4pt]
  0 & 1
  \end{bmatrix}
  }_{B}
  \tilde{w}(t)
  $$

  where $\tilde{w}(t)$ is a **2×1 vector** of random values (process noise).

- One possible **output equation** is:

  $$
  z(t) =
  \underbrace{
  \begin{bmatrix}
  1 & 0 & 0 & 0 \\[4pt]
  0 & 0 & 1 & 0
  \end{bmatrix}
  }_{C}
  x(t)
  + v(t)
  $$

  where $v(t)$ is sensor noise added to the measured position.

---

✅ Notes:
- The NCP model describes **static or nearly static** targets.
- The NCV model describes **moving targets with approximately constant velocity**.
- These are fundamental target-dynamics models used in **Kalman filter-based tracking**.


### 🧩 Example: The Coordinated-Turn (CT) Model

- Now, consider an object moving in a 2-D plane with **constant speed** and **angular rate** $\Omega$,  
  where $\Omega > 0$ indicates **counter-clockwise** motion and $\Omega < 0$ indicates **clockwise** motion.

  $$
  \dot{\xi}(t) = -\Omega \dot{\eta}(t), \quad
  \ddot{\eta}(t) = \Omega \dot{\xi}(t)
  $$

- We keep the same model state and modify the **state equation** to be:

  $$
  x(t) =
  \begin{bmatrix}
  \xi(t) \\[4pt]
  \dot{\xi}(t) \\[4pt]
  \eta(t) \\[4pt]
  \dot{\eta}(t)
  \end{bmatrix},
  \quad
  \dot{x}(t) =
  \underbrace{
  \begin{bmatrix}
  0 & 1 & 0 & 0 \\[4pt]
  0 & 0 & 0 & -\Omega \\[4pt]
  0 & 0 & 0 & 1 \\[4pt]
  0 & \Omega & 0 & 0
  \end{bmatrix}
  }_{A}
  x(t)
  +
  \underbrace{
  \begin{bmatrix}
  0 & 0 \\[4pt]
  1 & 0 \\[4pt]
  0 & 0 \\[4pt]
  0 & 1
  \end{bmatrix}
  }_{B}
  \tilde{w}(t)
  $$

  where $\tilde{w}(t)$ is a 2×1 vector of **random process-noise** values.

- One possible **output equation** is:

  $$
  z(t) =
  \underbrace{
  \begin{bmatrix}
  1 & 0 & 0 & 0 \\[4pt]
  0 & 0 & 1 & 0
  \end{bmatrix}
  }_{C}
  x(t)
  + v(t)
  $$

  where $v(t)$ is the sensor noise.

---

### 🧩 Simulating Continuous-Time Systems

- **Octave** has two functions in its *control* package that help simulate continuous-time state-space models easily:
  1. `ss()` — creates a state-space object from matrices $A$, $B$, $C$, and $D$.
  2. `lsim()` — simulates the state-space object for a given input and initial state.

- To simulate a model:
  - First define the matrices $A$, $B$, $C$, and $D$ in the Octave workspace.
  - Create a state-space model using `ss`.
  - Define a time vector and input sequence in the workspace.
  - Simulate the model for that input sequence using `lsim`.
  - The output of the simulation is the signal $z(t)$.

- The following slides show example simulations.

- Let’s first consider simulating an **NCP model**.

- The figure on the right (from Octave) shows the **trajectory of NCP model simulation**:

  C = eye(2); Dv = zeros(2);

  z = lsim(ncp,w',t)' + v;
  plot(z(1,:), z(2,:));
